```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
models = ['pcv2-small', 'pcv2-medium', 'pcv2-large']
total_throughput = [160422.15, 57107.02, 32872.07]
mlm_accuracy = [0.598, 0.641, 0.657]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize = (10, 5))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # Blue, Orange, Green
bars1 = ax1.bar(models, total_throughput, color=colors, alpha=0.8)
ax1.set_ylabel('Total Throughput (tokens/sec)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Model', fontsize=12, fontweight='bold')
ax1.grid(axis='y', alpha=0.3)

for bar, value in zip(bars1, total_throughput):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + height*0.01,
             f'{value:,.0f}', ha='center', va='bottom', fontweight='bold')

# Plot 2: Per-Sample Throughput  
bars2 = ax2.bar(models, mlm_accuracy, color=colors, alpha=0.8)
ax2.set_ylabel('MLM accuracy (15% masking rate)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Model', fontsize=12, fontweight='bold')
ax2.grid(axis='y', alpha=0.3)

for bar, value in zip(bars2, mlm_accuracy):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + height*0.01,
             f'{value:,.3f}', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('Figure-S1.pdf', format = 'pdf', bbox_inches = 'tight')
plt.show()
```
