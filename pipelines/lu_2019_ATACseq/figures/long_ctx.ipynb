```python
import matplotlib.pyplot as plt
```

```python
# Data
context = [600, 600+500*2, 600+1000*2, 600+2000*2]  # bp
hordeum_vulgare = [0.397604619, 0.4286865, 0.4759922, 0.4788944]
zea_mays = [0.586669524, 0.6767594, 0.7219182, 0.7110179]
```

```python
# Plot
plt.figure(figsize=(5, 4))

plt.plot(context, hordeum_vulgare, marker='o', color='orange', linewidth=2, markersize=6, label='Hordeum vulgare')
plt.plot(context, zea_mays, marker='o', color='green', linewidth=2, markersize=6,label='Zea mays')
plt.grid(True, alpha=0.3)
plt.xlabel('Context window length (bp)')
plt.ylabel('AUROC')
plt.ylim(0.3, 0.8)
plt.xticks(context)
plt.legend(frameon=False)
plt.tight_layout()
plt.savefig('performance_vs_context_window.pdf', format='pdf', bbox_inches='tight')
plt.show()
```
