```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
import pysam
from tqdm import tqdm
```

```python
peakDir = '/workdir/jz963/utils/plantcad2/results/lu_2019_ATACseq/peaks/'
fnames = os.listdir(peakDir)
```

```python
plt.figure(figsize=(15, 10))
plt.subplots_adjust(hspace=0.5)
for i, fname in enumerate(fnames):
    plt.subplot(3, 4, i+1) 
    curBED = pd.read_csv(f'{peakDir}/{fname}', header=None, sep='\t')
    Len = curBED[2] - curBED[1]
    plt.hist(Len, bins=60, alpha=0.7, color='skyblue')
    title = fname.replace('.bed', '')
    plt.title(f'{title}')
    plt.xlabel('Accessible regions length')
    plt.ylabel('Frequency')
    plt.xlim(0, 2000)

plt.tight_layout()
plt.show()
```
