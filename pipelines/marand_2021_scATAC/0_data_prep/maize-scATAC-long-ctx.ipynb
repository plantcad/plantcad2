```python
import pyfaidx
import pandas as pd
from tqdm import tqdm
import os
```

```python
# Parameters
flanks = [500, 1000, 2000]
datasets = ['train', 'valid', 'test']
genome = pyfaidx.Fasta('../../../results/marand_2021_scATAC/AGPv4.fa')

for flank in flanks:
    expected_length = 600 + 2 * flank
    
    for dataset in datasets:
        print(f"Processing {dataset}.tsv with flank={flank}")
        
        # Read dataset
        df = pd.read_csv(f"../../../results/PlantCAD2_tasks/maize_cell_type_accessible_c600/{dataset}.tsv", sep="\t")
        
        sequences = []
        for index, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing sequences"):
            chrom = str(row['Chr'])
            start = int(row['Start']) - flank
            end = int(row['End']) + flank
            
            # Adjust to chromosome length
            if start < 0:
                start = 0
            if end > len(genome[chrom]):
                end = len(genome[chrom])
            
            # Extract sequence
            seq = genome[chrom][start:end]
            sequences.append(seq)
        
        # Add sequences to dataframe
        df['Seq'] = sequences
        df = df[df['Seq'].str.len() == expected_length]
        
        # Save with flank in filename
        output_file = f"../../../results/PlantCAD2_tasks/maize_cell_type_accessible_c600/{dataset}_flank_{flank}.tsv"
        df.to_csv(output_file, sep="\t", index=False)
        print(f"Saved: {output_file}")
```
