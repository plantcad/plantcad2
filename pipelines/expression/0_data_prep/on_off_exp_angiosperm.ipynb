```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
train = pd.read_csv('../../results/PlantCAD2_tasks/max_exp_angiosperm/train.tsv', sep='\t')
valid = pd.read_csv('../../results/PlantCAD2_tasks/max_exp_angiosperm/valid.tsv', sep='\t')
ath = pd.concat((train, valid), axis = 0)
```

```python
off_cutoff = np.log10(2)
```

```python
ath['label'] = -1
ath.loc[ath['Label'] > off_cutoff, 'label'] = 1
ath.loc[ath['Label'] <= off_cutoff, 'label'] = 0
ath = ath[~(ath['label'] == -1)]
```

```python
train = ath[ath['group_for_cross_validation'] != 5]
valid = ath[ath['group_for_cross_validation'] == 5]
train.shape, valid.shape
```

```python
outputDir = '../../results/PlantCAD2_tasks/on_off_exp_angiosperm/'
os.makedirs(outputDir, exist_ok=True)
```

```python
train = train.drop(['Label'], axis = 1)
valid = valid.drop(['Label'], axis = 1)
train = train.rename(columns = {'label':'Label'})
valid = valid.rename(columns = {'label':'Label'})
```

```python
train.to_csv(f'{outputDir}train.tsv', sep='\t', index=False)
valid.to_csv(f'{outputDir}valid.tsv', sep='\t', index=False)
```

```python
folder_path = "../../results/PlantCAD2_tasks/max_exp_angiosperm/"
files = os.listdir(folder_path)[2:]
```

```python
for fname in files:
    curPath = f'{folder_path}{fname}'
    test = pd.read_csv(curPath, sep='\t')
    test['label'] = -1
    test.loc[test['Label'] > off_cutoff, 'label'] = 1
    test.loc[test['Label'] <= off_cutoff, 'label'] = 0
    test = test[~(test['label'] == -1)]
    test = test.drop(['Label'], axis = 1)
    test = test.rename(columns = {'label':'Label'})
    test.to_csv(f'{outputDir}{fname}', sep='\t', index=False)
    print(fname, test['Label'].value_counts())
```
